## Hallazgos de la exploración inicial

- Dataset con 10,763 filas y 23 columnas.
- Sin filas duplicadas.
- Nulos relevantes en: `promedio_ingresos_datacredito` (~27%) y `tendencia_ingresos` (~27%). 
  Nulos menores en `saldo_mora`, `saldo_total`, `saldo_principal`, `saldo_mora_codeudor`, `puntaje_datacredito`.
- Valores fuera de rango detectados:
  - `edad_cliente`: máximo de 123 años (dato inválido).
  - `puntaje` y `puntaje_datacredito`: valores negativos (inválidos para un score).
  - `salario_cliente`: mínimo de $0 y máximo de $22.000 millones (posibles errores de digitación).
  - `total_otros_prestamos`: máximo desproporcionado respecto a la mediana.
- Variable objetivo `Pago_atiempo` está desbalanceada: ~95% clase 1, ~5% clase 0. 
  Se deberá tratar este desbalance al momento de entrenar los modelos.
